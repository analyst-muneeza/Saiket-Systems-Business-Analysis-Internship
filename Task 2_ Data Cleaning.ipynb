# Task 2: Data Cleaning

# Objective – Task 2

The objective of this task is to clean and prepare the Telco Customer Churn dataset for analysis by handling missing values, removing duplicate records, and standardizing column names. Proper data cleaning ensures accuracy, consistency, and reliability for further analysis and visualization.

```python
# Import pandas
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("Telco_Customer_Churn_Dataset.csv")
```

```python
# Check missing values in each column
df.isnull().sum()
```

```python
# Total missing values in dataset
df.isnull().sum().sum()
```

```python
# Fill missing numerical values with mean
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
# Fill categorical missing values with mode
for col in df.select_dtypes(include='object').columns:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
# Count duplicate rows
df.duplicated().sum()
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```

```python
# Check duplicates again
df.duplicated().sum()
```

```python
# Display column names
df.columns
```

```python
# Convert column names to lowercase
# Replace spaces with underscores

df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
# Check updated column names
df.columns
```

```python
# Rows and columns after cleaning
df.shape
```

```python
# Final cleaned dataset info
df.info()
```

```python
# Save cleaned dataset
df.to_csv("cleaned_telco_customer_churn.csv", index=False)
```

# Conclusion – Task 2

In this task, the dataset was cleaned and prepared for analysis by handling missing values, removing duplicate records, and standardizing column names. These preprocessing steps improved the overall quality and consistency of the dataset. The cleaned dataset is now ready for exploratory data analysis and visualization in the next tasks.
