# Task 3: Exploratory Data Analysis (EDA)

# Objective

The objective of this task is to analyze and understand patterns, trends, and distributions in the Telco Customer Churn dataset using statistical analysis and data visualization techniques. This includes generating summary statistics, visualizing numerical data, and analyzing customer churn behavior.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("Telco_Customer_Churn_Dataset.csv")
```

```python
# Display first 5 rows
df.head()
```

```python
# Generate statistical summary
df.describe()
```

```python
# Find mean of Monthly Charges
df['MonthlyCharges'].mean()
```

```python
# Find median of Monthly Charges
df['MonthlyCharges'].median()
```

```python
# Find mode of Monthly Charges
df['MonthlyCharges'].mode()
```

```python
# Histogram for tenure column
plt.figure(figsize=(8,5))
plt.hist(df['tenure'], bins=20, edgecolor = 'black')

# Add title and labels
plt.title("Distribution of Tenure")
plt.xlabel("Tenure")
plt.ylabel("Frequency")

# Show plot
plt.show()
```

```python
# Histogram for Monthly Charges
plt.figure(figsize=(8,5))
plt.hist(df['MonthlyCharges'], bins=20, edgecolor = 'black')

# Add title and labels
plt.title("Distribution of Monthly Charges")
plt.xlabel("Monthly Charges")
plt.ylabel("Frequency")

# Show plot
plt.show()
```

```python
# Box plot for Monthly Charges
plt.figure(figsize=(8,5))

sns.boxplot(x=df['MonthlyCharges'])

# Add title
plt.title("Box Plot of Monthly Charges")

# Show plot
plt.show()
```

```python
# Box plot for tenure
plt.figure(figsize=(8,5))

sns.boxplot(x=df['tenure'])

# Add title
plt.title("Box Plot of Tenure")

# Show plot
plt.show()
```

```python
# Count churn and non-churn customers
df['Churn'].value_counts()
```

```python
# Pie chart for churn distribution
churn_counts = df['Churn'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(churn_counts,
        labels=churn_counts.index,
        autopct='%1.1f%%')

# Add title
plt.title("Churn vs Non-Churn Customers")

# Show chart
plt.show()
```

```python
# Count plot of churn customers
plt.figure(figsize=(6,5))

sns.countplot(x='Churn', data=df)

# Add title
plt.title("Customer Churn Count")

# Show plot
plt.show()
```

# Conclusion

In this task, Exploratory Data Analysis (EDA) was performed on the Telco Customer Churn dataset to understand customer behavior and data patterns. Summary statistics such as mean, median, and mode were calculated for numerical features. Histograms and box plots were created to analyze distributions and detect outliers. Churn analysis helped identify the proportion of churned and non-churned customers. These insights provide a better understanding of the dataset and support further business analysis and predictive modeling.
