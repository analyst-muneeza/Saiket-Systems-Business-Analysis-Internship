# Task 5: Advanced Analysis

# Objective

The objective of this task is to perform advanced analysis on the Telco Customer Churn dataset by examining customer behavior across different tenure groups, demographics, payment methods, and contract types. The analysis aims to identify factors influencing churn and provide deeper business insights through data visualization and statistical comparisons.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("Telco_Customer_Churn_Dataset.csv")
```

```python
# Create tenure categories

df['tenure_group'] = pd.cut(df['tenure'],
                            bins=[0,12,36,72],
                            labels=['0-12 Months',
                                    '13-36 Months',
                                    '37+ Months'])
```

```python
# Calculate statistics for Monthly Charges by tenure group

df.groupby('tenure_group')['MonthlyCharges'].agg(
    ['mean','median','min','max']
)
```

```python
# Churn rate across tenure groups

pd.crosstab(df['tenure_group'],
            df['Churn'],
            normalize='index')*100
```

```python
# Visualize churn by tenure group

sns.countplot(x='tenure_group',
              hue='Churn',
              data=df)

plt.title("Churn by Tenure Group")
plt.show()
```

```python
# Analyze churn based on gender

pd.crosstab(df['gender'],
            df['Churn'])
```

```python
# Visualization

sns.countplot(x='gender',
              hue='Churn',
              data=df)

plt.title("Churn by Gender")
plt.show()
```

```python
# Churn analysis for senior citizens

pd.crosstab(df['SeniorCitizen'],
            df['Churn'])
```

```python
# Visualization

sns.countplot(x='SeniorCitizen',
              hue='Churn',
              data=df)

plt.title("Churn by Senior Citizen Status")
plt.show()
```

```python
# Analyze churn by payment method

pd.crosstab(df['PaymentMethod'],
            df['Churn'])
```

```python
# Visualization

plt.figure(figsize=(10,5))

sns.countplot(x='PaymentMethod',
              hue='Churn',
              data=df)

plt.xticks(rotation=45)
plt.title("Churn by Payment Method")

plt.show()
```

```python
# Analyze churn by contract type

pd.crosstab(df['Contract'],
            df['Churn'])
```

```python
# Visualization

sns.countplot(x='Contract',
              hue='Churn',
              data=df)

plt.title("Churn by Contract Type")

plt.show()
```

```python
# Average monthly charges by tenure group

avg_charges = df.groupby('tenure_group')['MonthlyCharges'].mean()

avg_charges.plot(kind='bar')

plt.title("Average Monthly Charges by Tenure Group")
plt.ylabel("Average Charges")

plt.show()
```

# Conclusion

In this task, advanced analysis was performed to identify factors affecting customer churn. Customers were grouped based on tenure, demographics, payment methods, and contract types to uncover meaningful patterns. The analysis revealed differences in churn behavior across customer segments and highlighted factors that may influence customer retention. These insights can help businesses develop targeted strategies to reduce churn and improve customer satisfaction.
