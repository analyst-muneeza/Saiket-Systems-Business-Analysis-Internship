# Task 4: Customer Segmentation Visualization

# Objective

The objective of this task is to visualize customer distribution based on tenure categories and compare average monthly charges across different customer groups using charts and graphs. This helps in understanding customer segments and identifying trends related to customer retention and spending behavior.

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("Telco_Customer_Churn_Dataset.csv")
```

```python
# Create customer tenure groups

df['tenure_group'] = pd.cut(df['tenure'],
                            bins=[0,12,36,72],
                            labels=['0-12 Months',
                                    '13-36 Months',
                                    '37+ Months'])
```

```python
# Count customers in each tenure group
df['tenure_group'].value_counts()
```

```python
# Customer distribution pie chart

tenure_counts = df['tenure_group'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(tenure_counts,
        labels=tenure_counts.index,
        autopct='%1.1f%%')

# Add title
plt.title("Customer Distribution by Tenure")

# Show chart
plt.show()
```

```python
# Donut chart for tenure distribution

plt.figure(figsize=(7,7))

plt.pie(tenure_counts,
        labels=tenure_counts.index,
        autopct='%1.1f%%')

# Create donut shape
centre_circle = plt.Circle((0,0),0.70,fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

# Add title
plt.title("Customer Segmentation Donut Chart")

# Display chart
plt.show()
```

```python
# Calculate average monthly charges

avg_charges = df.groupby('tenure_group')['MonthlyCharges'].mean()

avg_charges
```

```python
# Bar chart for average monthly charges

plt.figure(figsize=(8,5))

bars = plt.bar(avg_charges.index,
               avg_charges.values)

# Add title and labels
plt.title("Average Monthly Charges by Tenure Group")
plt.xlabel("Tenure Group")
plt.ylabel("Average Monthly Charges")

# Add annotations on bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             yval + 1,
             round(yval,2),
             ha='center')

# Show chart
plt.show()
```

# Conclusion

In this task, customer segmentation visualization was performed using tenure-based categories. Pie and donut charts were used to analyze customer distribution across different tenure groups. A clustered bar chart was created to compare average monthly charges among customer segments. The visualizations helped identify customer behavior patterns and spending trends based on customer tenure.
